# ENOE 2026 T1 — Verificación de llaves y construcción de un dataset unificado multinivel

**Contexto.** Insumo para un modelo de estimación de ingresos en un buró de crédito.
La ENOE se usa aquí como fuente auxiliar (no vinculable a nivel persona con datos crediticios),
en tres modos posibles: (a) corpus de entrenamiento de un modelo de ingreso sobre atributos
sociodemográficos, (b) generador de agregados geográficos para enriquecer *no-hits*,
(c) benchmark de calibración de distribuciones de ingreso.

---

## 1. Resultado de la verificación de unibilidad

**Sí es posible unir los 5 datasets.** El modelo entidad-relación oficial de INEGI
(`modelo_entidad_relacion/modelo_er_enoe_2026_1t.jpg`) declara esta jerarquía:

| Tabla | Nivel | Llave declarada por INEGI |
|---|---|---|
| `VIV` | Vivienda | `tipo, cve_ent, con, v_sel` |
| `HOG` | Hogar | `tipo, cve_ent, con, v_sel` + `n_hog, h_mud` |
| `SDEM` | Persona | ... + `n_ren` |
| `COE1` | Persona (15+) | ... + `n_ren` |
| `COE2` | Persona (15+) | ... + `n_ren` |

Las 5 tablas comparten además **todo el bloque muestral**: `cd_a, upm, d_sem, n_pro_viv,
n_ent, per, mes_cal, ur, cve_mun, cvegeo, fac_tri, fac_men`.

**Advertencia importante:** el diagrama E-R de INEGI omite `cd_a` y `mes_cal` de la llave, pero
`con` (control) sólo es único *dentro* de ciudad autorrepresentada, y el archivo trimestral
apila los 3 meses del trimestre. Por eso este notebook **no asume la llave: la prueba
empíricamente** contra varios conjuntos candidatos y elige el mínimo que resulta único
(sección 4). Si el diagrama fuera suficiente, la prueba lo confirmará; si no, verás los
duplicados explícitamente.

### Cardinalidades esperadas

```
VIV  1 ──< HOG  1 ──< SDEM  1 ──1 COE1
                        └───1 COE2
```

- `VIV → HOG`: 1:N (una vivienda puede alojar varios hogares; ver `viv.p3`).
- `HOG → SDEM`: 1:N. **No todos los hogares tienen personas**: `HOG` incluye no-respuesta
  (`r_def != 0`), `SDEM` sólo entrevistas resueltas. Esperar *anti-join* en ese sentido.
- `SDEM → COE1/COE2`: 1:1 **parcial**. El COE sólo se aplica a población de 15 años y más
  con entrevista completa. Un left-join desde `SDEM` dejará nulos en menores de 15 y en
  ausentes definitivos (`c_res == 2`).

### Diferencias de contenido relevantes

- `VIV` y `HOG` tienen el bloque geográfico fino (`cve_loc`, `cve_ageb`, `est`, `est_d_tri`,
  `est_d_men`, `t_loc_tri`); `COE1`/`COE2` **no** lo traen. `SDEM` sí. Por eso conviene usar
  `SDEM` como tabla base.
- La variable objetivo natural, **`ingocup` (ingreso mensual)**, vive en `SDEM`.
- El ingreso declarado crudo vive en `COE2`: `p6b1` (periodicidad) + `p6b2` (monto en pesos),
  `p6c` (múltiplos de salario mínimo, para quien no declara monto) y `p7gcan`
  (ingreso del segundo trabajo).

---

## 2. Estrategia del notebook

1. Carga robusta (todo como texto, para no perder ceros a la izquierda).
2. Diagnóstico de unicidad de llaves candidatas por tabla.
3. Diagnóstico de integridad referencial (anti-joins y cardinalidad real).
4. Merge en cascada con `validate=` e `indicator=` en cada paso.
5. Derivación de **agregados a nivel hogar** desde `SDEM`.
6. Derivación de **agregados geográficos ponderados** (`fac_tri`) a nivel entidad,
   municipio, ciudad autorrepresentada y estrato de diseño.
7. Ensamble final persona ← hogar ← vivienda ← agregados geo.
8. Validación reconstruyendo indicadores oficiales de la ENOE.
9. Exportación a Parquet/CSV + diccionario del dataset final.

## 3. Configuración y carga

````python
from __future__ import annotations

import unicodedata
import warnings
from pathlib import Path

import numpy as np
import pandas as pd

warnings.filterwarnings("ignore")
pd.set_option("display.max_columns", 250)
pd.set_option("display.width", 220)
pd.set_option("display.float_format", lambda v: f"{v:,.4f}")

print("pandas", pd.__version__)
````

````python
# ============================================================================
# CONFIGURACIÓN
# ============================================================================
# MODO = "local"  -> clon del repo. Requiere `git lfs install && git lfs pull`
#                    porque los .csv de conjunto_de_datos están en Git LFS.
# MODO = "github" -> lee directo del endpoint de media de GitHub (Colab-friendly,
#                    no requiere clonar ni git-lfs).
MODO = "local"

REPO_LOCAL  = Path("./datos_EI")
GH_MEDIA    = "https://media.githubusercontent.com/media/Ruhguevara/datos_EI/main"  # sirve LFS
GH_RAW      = "https://raw.githubusercontent.com/Ruhguevara/datos_EI/main"          # sirve texto plano

TRIMESTRE    = "2026_1t"
CARPETA_RAIZ = f"conjunto_de_datos_enoe_{TRIMESTRE}_csv"
TABLAS       = ["viv", "hog", "sdem", "coe1", "coe2"]

SALIDA = Path("./salida")
SALIDA.mkdir(exist_ok=True)


def _rel_datos(tabla: str) -> str:
    b = f"conjunto_de_datos_{tabla}_enoe_{TRIMESTRE}"
    return f"{CARPETA_RAIZ}/{b}/conjunto_de_datos/{b}.csv"


def _rel_dicc(tabla: str) -> str:
    b = f"conjunto_de_datos_{tabla}_enoe_{TRIMESTRE}"
    return f"{CARPETA_RAIZ}/{b}/diccionario_de_datos/diccionario_datos_{tabla}_enoe_{TRIMESTRE}.csv"


def _rel_catalogo(tabla: str, cat: str) -> str:
    b = f"conjunto_de_datos_{tabla}_enoe_{TRIMESTRE}"
    return f"{CARPETA_RAIZ}/{b}/catalogos/{cat}.csv"


def ruta(rel: str, lfs: bool) -> str:
    """Resuelve una ruta relativa del repo a ruta local o URL."""
    if MODO == "local":
        return str(REPO_LOCAL / rel)
    return f"{GH_MEDIA}/{rel}" if lfs else f"{GH_RAW}/{rel}"
````

````python
# ============================================================================
# LECTURA ROBUSTA
# ============================================================================
# Regla dura: TODO se lee como texto. Las llaves de la ENOE (cve_ent="01",
# con="00123", v_sel="01", n_ent="01") tienen ceros a la izquierda. Si pandas
# infiere int, los joins fallan en silencio. La conversión numérica es explícita
# y posterior, guiada por el diccionario.

_ENCODINGS = ("utf-8-sig", "latin-1", "cp1252")


def _sin_acentos(s: str) -> str:
    s = unicodedata.normalize("NFKD", str(s))
    return "".join(c for c in s if not unicodedata.combining(c))


def _leer_csv(path: str, **kw) -> pd.DataFrame:
    ultimo = None
    for enc in _ENCODINGS:
        try:
            return pd.read_csv(path, encoding=enc, dtype=str, low_memory=False, **kw)
        except UnicodeDecodeError as e:
            ultimo = e
    raise ultimo


def leer_tabla(tabla: str) -> pd.DataFrame:
    df = _leer_csv(ruta(_rel_datos(tabla), lfs=True))
    df.columns = [c.strip().lower() for c in df.columns]
    # Blindaje: si el archivo sigue siendo un puntero de Git LFS, avisar claro.
    if df.shape[1] <= 2 and "version https" in str(df.columns[0]) + str(df.iloc[0, 0] if len(df) else ""):
        raise RuntimeError(
            f"'{tabla}' es un puntero de Git LFS, no el CSV real.\n"
            "Ejecuta `git lfs install && git lfs pull` en el clon, o usa MODO='github'."
        )
    return df


def leer_diccionario(tabla: str) -> pd.DataFrame:
    d = _leer_csv(ruta(_rel_dicc(tabla), lfs=False))
    d.columns = [_sin_acentos(c).strip().lower() for c in d.columns]
    d = d.rename(columns={
        "nemonico": "variable",
        "nombre_campo": "descripcion",
        "longitud": "longitud",
        "tipo": "tipo_dato",
        "catalogo": "catalogo",
    })
    d["variable"] = d["variable"].astype("string").str.strip().str.lower()
    d["tabla"] = tabla
    cols = ["tabla", "variable", "descripcion", "tipo_dato", "longitud", "catalogo"]
    return d[[c for c in cols if c in d.columns]]


def leer_catalogo(tabla: str, cat: str) -> dict:
    """Devuelve {clave: descripción} de un catálogo de INEGI."""
    c = _leer_csv(ruta(_rel_catalogo(tabla, cat), lfs=False))
    c.columns = [x.strip().lower() for x in c.columns]
    return dict(zip(c["cve"].str.strip(), c["descrip"].str.strip()))
````

````python
# ============================================================================
# CARGA
# ============================================================================
raw = {t: leer_tabla(t) for t in TABLAS}
dicc = pd.concat([leer_diccionario(t) for t in TABLAS], ignore_index=True)

resumen = pd.DataFrame(
    [{"tabla": t, "filas": len(df), "columnas": df.shape[1],
      "MB_en_memoria": round(df.memory_usage(deep=True).sum() / 1024**2, 1)}
     for t, df in raw.items()]
)
display(resumen)
print("\nDiccionario consolidado:", dicc.shape)
````

## 4. Verificación de llaves

Se prueban tres conjuntos candidatos por nivel:

| Candidato | Contenido | Racional |
|---|---|---|
| `INEGI_ER` | lo que dice el diagrama E-R | verificar la documentación oficial |
| `MINIMA`   | E-R + `cd_a` + `mes_cal` | `con` es único dentro de ciudad; el archivo apila 3 meses |
| `COMPLETA` | + `upm`, `d_sem`, `n_pro_viv`, `n_ent`, `per` | superconjunto muestral, redundante pero a prueba de balas |

La llave que se use en producción es la **mínima que resulte única en las 5 tablas**.

````python
# ============================================================================
# DEFINICIÓN DE LLAVES CANDIDATAS
# ============================================================================
K_ER_VIV   = ["tipo", "cve_ent", "con", "v_sel"]
K_MIN_VIV  = ["tipo", "mes_cal", "cd_a", "cve_ent", "con", "v_sel"]
K_FULL_VIV = ["tipo", "mes_cal", "cd_a", "cve_ent", "con", "upm", "d_sem",
              "n_pro_viv", "v_sel", "n_ent", "per"]

CANDIDATAS = {
    "INEGI_ER": {"viv": K_ER_VIV,
                 "hog": K_ER_VIV + ["n_hog", "h_mud"],
                 "per": K_ER_VIV + ["n_hog", "h_mud", "n_ren"]},
    "MINIMA":   {"viv": K_MIN_VIV,
                 "hog": K_MIN_VIV + ["n_hog", "h_mud"],
                 "per": K_MIN_VIV + ["n_hog", "h_mud", "n_ren"]},
    "COMPLETA": {"viv": K_FULL_VIV,
                 "hog": K_FULL_VIV + ["n_hog", "h_mud"],
                 "per": K_FULL_VIV + ["n_hog", "h_mud", "n_ren"]},
}

NIVEL_DE_TABLA = {"viv": "viv", "hog": "hog", "sdem": "per", "coe1": "per", "coe2": "per"}

# Anchos oficiales por variable (para normalizar ceros a la izquierda)
ANCHOS = (
    dicc.assign(_l=pd.to_numeric(dicc["longitud"], errors="coerce"))
        .groupby("variable")["_l"].max()
        .dropna().astype(int).to_dict()
)
print({k: ANCHOS[k] for k in
       ["tipo", "mes_cal", "cd_a", "cve_ent", "con", "upm", "d_sem", "n_pro_viv",
        "v_sel", "n_ent", "per", "n_hog", "h_mud", "n_ren", "cve_mun", "cvegeo"]})
````

````python
# ============================================================================
# NORMALIZACIÓN DE LLAVES
# ============================================================================
def normalizar_llaves(df: pd.DataFrame, columnas) -> pd.DataFrame:
    """Recorta espacios y rellena con ceros a la izquierda al ancho oficial.

    Esto neutraliza la diferencia entre exportaciones que guardan n_ent='1'
    y otras que guardan '01', que es la causa #1 de joins vacíos con ENOE.
    """
    df = df.copy()
    for c in columnas:
        if c not in df.columns:
            continue
        s = df[c].astype("string").str.strip()
        w = ANCHOS.get(c)
        if w:
            s = s.str.zfill(w)
        df[c] = s
    return df


TODAS_LAS_LLAVES = sorted({c for cand in CANDIDATAS.values()
                             for llave in cand.values() for c in llave})

norm = {t: normalizar_llaves(df, TODAS_LAS_LLAVES) for t, df in raw.items()}
print("Columnas de llave normalizadas:", TODAS_LAS_LLAVES)
````

````python
# ============================================================================
# PRUEBA DE UNICIDAD
# ============================================================================
def diagnostico_llave(df: pd.DataFrame, llave: list, tabla: str, nombre: str) -> dict:
    faltan = [c for c in llave if c not in df.columns]
    if faltan:
        return {"tabla": tabla, "candidata": nombre, "estado": f"FALTAN: {faltan}",
                "filas": len(df), "unicas": np.nan, "duplicadas": np.nan,
                "pct_dup": np.nan, "nulos_en_llave": np.nan, "es_unica": False}
    sub = df[llave]
    n = len(sub)
    u = int(len(sub.drop_duplicates()))
    nulos = int(sub.isna().any(axis=1).sum())
    return {"tabla": tabla, "candidata": nombre, "estado": "ok",
            "filas": n, "unicas": u, "duplicadas": n - u,
            "pct_dup": round(100 * (n - u) / n, 4) if n else np.nan,
            "nulos_en_llave": nulos, "es_unica": (u == n and nulos == 0)}


filas = []
for nombre, cand in CANDIDATAS.items():
    for tabla, df in norm.items():
        filas.append(diagnostico_llave(df, cand[NIVEL_DE_TABLA[tabla]], tabla, nombre))

unicidad = pd.DataFrame(filas)
display(unicidad.pivot(index="tabla", columns="candidata", values="pct_dup")
                .rename_axis(columns="% filas duplicadas"))
display(unicidad)
````

````python
# ============================================================================
# SELECCIÓN AUTOMÁTICA DE LA LLAVE MÍNIMA VÁLIDA
# ============================================================================
orden = ["INEGI_ER", "MINIMA", "COMPLETA"]
elegida = None
for nombre in orden:
    bloque = unicidad[unicidad["candidata"] == nombre]
    if bloque["es_unica"].all():
        elegida = nombre
        break

if elegida is None:
    raise AssertionError(
        "Ninguna llave candidata es única en las 5 tablas. "
        "Revisa el DataFrame `unicidad` antes de continuar."
    )

K_VIV = CANDIDATAS[elegida]["viv"]
K_HOG = CANDIDATAS[elegida]["hog"]
K_PER = CANDIDATAS[elegida]["per"]

print(f"Llave elegida: {elegida}")
print(f"  K_VIV ({len(K_VIV)}): {K_VIV}")
print(f"  K_HOG ({len(K_HOG)}): {K_HOG}")
print(f"  K_PER ({len(K_PER)}): {K_PER}")
````

### 4.1 Integridad referencial

Antes de unir, hay que saber cuántas llaves de cada lado quedan huérfanas. Un
`left join` silencioso puede ocultar que el 30% de un archivo no cruza.

````python
# ============================================================================
# ANTI-JOINS Y CARDINALIDAD REAL
# ============================================================================
def integridad(izq: pd.DataFrame, der: pd.DataFrame, llave: list,
               n_izq: str, n_der: str) -> dict:
    a = izq[llave].drop_duplicates()
    b = der[llave].drop_duplicates()
    m = a.merge(b, on=llave, how="outer", indicator=True)
    cnt = m["_merge"].value_counts()
    solo_a = int(cnt.get("left_only", 0))
    solo_b = int(cnt.get("right_only", 0))
    ambos  = int(cnt.get("both", 0))
    # cardinalidad: filas por llave en cada lado
    max_izq = int(izq.groupby(llave, dropna=False).size().max())
    max_der = int(der.groupby(llave, dropna=False).size().max())
    return {
        "relacion": f"{n_izq} -> {n_der}",
        "llaves_izq": len(a), "llaves_der": len(b),
        "cruzan": ambos,
        f"solo_{n_izq}": solo_a, f"solo_{n_der}": solo_b,
        "%_izq_cruza": round(100 * ambos / len(a), 2) if len(a) else np.nan,
        "%_der_cruza": round(100 * ambos / len(b), 2) if len(b) else np.nan,
        "cardinalidad": f"{max_izq}:{max_der}",
    }


pruebas = [
    integridad(norm["viv"],  norm["hog"],  K_VIV, "viv",  "hog"),
    integridad(norm["hog"],  norm["sdem"], K_HOG, "hog",  "sdem"),
    integridad(norm["sdem"], norm["coe1"], K_PER, "sdem", "coe1"),
    integridad(norm["sdem"], norm["coe2"], K_PER, "sdem", "coe2"),
    integridad(norm["coe1"], norm["coe2"], K_PER, "coe1", "coe2"),
]
ref = pd.DataFrame(pruebas)
display(ref.set_index("relacion"))
````

````python
# ============================================================================
# ¿QUIÉN NO ESTÁ EN EL COE Y POR QUÉ?
# ============================================================================
# Hipótesis a validar: el COE sólo cubre 15+ con entrevista resuelta.
sd = norm["sdem"].copy()
sd["_eda"] = pd.to_numeric(sd["eda"], errors="coerce").where(lambda s: s < 98)

en_coe1 = sd[K_PER].merge(norm["coe1"][K_PER].assign(_hit=1), on=K_PER, how="left")["_hit"].notna()
sd["_en_coe1"] = en_coe1.values

tabla = (sd.assign(grupo=np.select(
            [sd["_eda"].lt(15), sd["_eda"].ge(15)],
            ["menor de 15", "15 y más"], default="edad no especificada"))
           .groupby(["grupo", "c_res"], dropna=False)["_en_coe1"]
           .agg(personas="size", con_coe1="sum")
           .assign(pct_cobertura=lambda d: (100 * d["con_coe1"] / d["personas"]).round(2))
           .reset_index())
display(tabla)
print("\nc_res: 1=residente habitual, 2=ausente definitivo, 3=nuevo residente")
````

## 5. Tipado y limpieza

El diccionario marca cada variable como `C` (carácter) o `N` (numérica). Además hay
variables `C` que son numéricas de facto (`eda`, `p6b2`, `p7gcan`, `p3q`, ...) y códigos
centinela de no-respuesta que hay que convertir a `NaN` **antes** de cualquier promedio.

````python
# ============================================================================
# CENTINELAS DE NO RESPUESTA (según catálogos de INEGI)
# ============================================================================
CENTINELAS = {
    "eda":       [98, 99],        # años no especificados
    "anios_esc": [99],            # escolaridad no especificada; 0 = "no aplica"
    "hrsocup":   [999],
    "ingocup":   [999999],
    "ing_x_hrs": [999999],
    "p6b2":      [999998, 999999],
    "p7gcan":    [999998, 999999],
    "p9mcan":    [999998, 999999],
    "p3q":       [98, 99],
    "n_hij":     [98, 99],
}

# Variables tipo C que en realidad son numéricas
EXTRA_NUMERICAS = {
    "eda", "n_hij", "nac_anio", "anios_esc", "p3q", "p3g_tot", "p6b2", "p7gcan",
    "p9mcan", "p5c_thrs", "p5e_thrs", "p5c_tdia", "p5e_tdia", "p7f_horas",
    "p7f_dias", "p1", "p3", "p4_2", "inf",
}


def tipar(df: pd.DataFrame, tabla: str) -> pd.DataFrame:
    """Convierte a numérico lo que el diccionario marca como N + extras conocidas."""
    d = dicc[dicc["tabla"] == tabla]
    num = set(d.loc[d["tipo_dato"].str.strip().eq("N"), "variable"]) | EXTRA_NUMERICAS
    num = {c for c in num if c in df.columns} - set(TODAS_LAS_LLAVES)

    out = df.copy()
    for c in sorted(num):
        s = pd.to_numeric(out[c], errors="coerce")
        for sent in CENTINELAS.get(c, []):
            s = s.mask(s.eq(sent))
        out[c] = s
    return out


tip = {t: tipar(norm[t], t) for t in TABLAS}

chk = tip["sdem"][["eda", "anios_esc", "hrsocup", "ingocup", "ing_x_hrs", "fac_tri"]].describe().T
display(chk)
````

## 6. Merge en cascada

Base = `SDEM` (persona). Es la única tabla que tiene simultáneamente el bloque geográfico
fino, la variable objetivo `ingocup` y todos los clasificadores derivados de INEGI.

Cada paso usa `validate=` (falla ruidosamente si la cardinalidad no es la esperada) e
`indicator=` (deja rastro de qué cruzó).

````python
# ============================================================================
# SELECCIÓN DE COLUMNAS POR TABLA
# ============================================================================
# Columnas administrativas/duplicadas que no queremos arrastrar desde las tablas
# secundarias (ya vienen en SDEM o en la llave).
COMUNES_A_DESCARTAR = {"ur", "fac_tri", "fac_men", "cve_mun", "cvegeo", "eda",
                       "n_inf", "est", "est_d_tri", "est_d_men", "cve_loc",
                       "cve_ageb", "t_loc_tri", "t_loc_men"}


def cols_utiles(tabla: str, llave: list, prefijo: str | None = None):
    df = tip[tabla]
    fuera = set(llave) | COMUNES_A_DESCARTAR
    cols = [c for c in df.columns if c not in fuera]
    sub = df[llave + cols].copy()
    if prefijo:
        sub = sub.rename(columns={c: f"{prefijo}{c}" for c in cols})
    return sub


def merge_seguro(izq, der, llave, como, etiqueta, validate):
    antes = len(izq)
    out = izq.merge(der, on=llave, how=como, validate=validate,
                    indicator=f"_m_{etiqueta}")
    assert len(out) == antes, (
        f"[{etiqueta}] el merge cambió el número de filas: {antes} -> {len(out)}"
    )
    vc = out[f"_m_{etiqueta}"].value_counts()
    pct = 100 * vc.get("both", 0) / len(out)
    print(f"{etiqueta:>6s}: {vc.get('both', 0):>8,} de {len(out):>8,} filas cruzan ({pct:5.2f}%)")
    out[f"_m_{etiqueta}"] = out[f"_m_{etiqueta}"].eq("both")
    return out
````

````python
# ============================================================================
# CASCADA: SDEM  <- COE1  <- COE2  <- HOG  <- VIV
# ============================================================================
base = tip["sdem"].copy()
print(f"base (SDEM): {len(base):,} personas\n")

base = merge_seguro(base, cols_utiles("coe1", K_PER, "c1_"), K_PER,
                    "left", "coe1", validate="one_to_one")
base = merge_seguro(base, cols_utiles("coe2", K_PER, "c2_"), K_PER,
                    "left", "coe2", validate="one_to_one")
base = merge_seguro(base, cols_utiles("hog", K_HOG, "hg_"), K_HOG,
                    "left", "hog",  validate="many_to_one")
base = merge_seguro(base, cols_utiles("viv", K_VIV, "vv_"), K_VIV,
                    "left", "viv",  validate="many_to_one")

print(f"\nResultado: {base.shape[0]:,} filas x {base.shape[1]:,} columnas")
````

## 7. Nivel hogar: agregados derivados

Para un modelo de ingreso en contexto crediticio, los agregados del hogar suelen tener
más señal que los atributos individuales aislados: ingreso total del hogar, número de
perceptores, razón de dependencia y perfil del jefe de hogar.

Se calculan sobre residentes habituales (`c_res != 2`) para no contar ausentes definitivos.

````python
# ============================================================================
# AGREGADOS A NIVEL HOGAR
# ============================================================================
def agregados_hogar(sdem: pd.DataFrame, llave_hog: list) -> pd.DataFrame:
    d = sdem[sdem["c_res"].ne("2")].copy()

    d["_ocupado"]     = d["clase2"].eq(1)
    d["_desocupado"]  = d["clase2"].eq(2)
    d["_pea"]         = d["clase1"].eq(1)
    d["_perceptor"]   = d["ingocup"].gt(0)
    d["_informal"]    = d["emp_ppal"].eq(1)
    d["_con_segsoc"]  = d["seg_soc"].eq(1)
    d["_menor15"]     = d["eda"].lt(15)
    d["_65mas"]       = d["eda"].ge(65)
    d["_edad_activa"] = d["eda"].between(15, 64)

    g = d.groupby(llave_hog, dropna=False, observed=True)
    agg = g.agg(
        hog_n_personas    = ("n_ren",      "size"),
        hog_n_menor15     = ("_menor15",   "sum"),
        hog_n_65mas       = ("_65mas",     "sum"),
        hog_n_edad_activa = ("_edad_activa", "sum"),
        hog_n_pea         = ("_pea",       "sum"),
        hog_n_ocupados    = ("_ocupado",   "sum"),
        hog_n_desocupados = ("_desocupado","sum"),
        hog_n_perceptores = ("_perceptor", "sum"),
        hog_n_informales  = ("_informal",  "sum"),
        hog_n_con_segsoc  = ("_con_segsoc","sum"),
        hog_ing_total     = ("ingocup",    "sum"),
        hog_ing_max       = ("ingocup",    "max"),
        hog_hrs_total     = ("hrsocup",    "sum"),
        hog_esc_max       = ("anios_esc",  "max"),
        hog_esc_media     = ("anios_esc",  "mean"),
        hog_eda_media     = ("eda",        "mean"),
    ).reset_index()

    # Perfil del jefe de hogar (par_c == '101')
    jefe = (d[d["par_c"].eq("101")]
              .sort_values(llave_hog + ["n_ren"])
              .drop_duplicates(subset=llave_hog, keep="first")
              [llave_hog + ["sex", "eda", "anios_esc", "niv_ins", "e_con",
                            "clase2", "pos_ocu", "emp_ppal", "seg_soc",
                            "ingocup", "hrsocup", "rama_est2", "scian"]]
              .rename(columns=lambda c: c if c in llave_hog else f"jefe_{c}"))

    out = agg.merge(jefe, on=llave_hog, how="left")

    # Derivadas
    out["hog_ing_pc"]          = out["hog_ing_total"] / out["hog_n_personas"].replace(0, np.nan)
    out["hog_ing_x_perceptor"] = out["hog_ing_total"] / out["hog_n_perceptores"].replace(0, np.nan)
    out["hog_razon_dep"]       = ((out["hog_n_menor15"] + out["hog_n_65mas"])
                                  / out["hog_n_edad_activa"].replace(0, np.nan))
    out["hog_tasa_ocup"]       = out["hog_n_ocupados"] / out["hog_n_pea"].replace(0, np.nan)
    out["hog_pct_informal"]    = out["hog_n_informales"] / out["hog_n_ocupados"].replace(0, np.nan)
    out["hog_pct_segsoc"]      = out["hog_n_con_segsoc"] / out["hog_n_ocupados"].replace(0, np.nan)
    out["hog_tiene_jefe"]      = out["jefe_sex"].notna()
    return out


hogar = agregados_hogar(tip["sdem"], K_HOG)
print(f"Hogares con al menos un residente: {len(hogar):,}")
display(hogar[["hog_n_personas", "hog_n_perceptores", "hog_ing_total",
               "hog_ing_pc", "hog_razon_dep"]].describe().T)
````

## 8. Niveles geográficos: agregados ponderados

Aquí está la parte directamente reutilizable para enriquecer *no-hits*: agregados
poblacionales por geografía, expandidos con `fac_tri`.

> **Advertencia de representatividad — leerla antes de usar el corte municipal.**
> La ENOE **no es representativa a nivel municipio** salvo en las ~39 ciudades
> autorrepresentadas (`cd_a`). Sus dominios de estimación oficiales son: nacional,
> estatal, las ciudades autorrepresentadas y los cortes urbano/rural por tamaño de
> localidad (`t_loc_tri`). Un promedio municipal calculado con 12 hogares tiene un
> error estándar enorme y, peor, sesgado por el diseño.
>
> Por eso se calculan cuatro niveles y se reporta el **tamaño de muestra efectivo**
> de cada celda. Para enriquecer *no-hits* recomiendo `cve_ent × t_loc_tri` o `cd_a`
> como nivel primario, y `cvegeo` (municipio) sólo con `n` suficiente y con
> encogimiento hacia la media estatal.

````python
# ============================================================================
# UTILIDADES DE ESTIMACIÓN PONDERADA
# ============================================================================
def media_pond(x: pd.Series, w: pd.Series) -> float:
    m = x.notna() & w.notna()
    if not m.any() or w[m].sum() == 0:
        return np.nan
    return float(np.average(x[m], weights=w[m]))


def cuantil_pond(x: pd.Series, w: pd.Series, q: float = 0.5) -> float:
    m = x.notna() & w.notna() & (w > 0)
    if not m.any():
        return np.nan
    xs, ws = x[m].to_numpy(float), w[m].to_numpy(float)
    o = np.argsort(xs)
    xs, ws = xs[o], ws[o]
    cw = np.cumsum(ws) - 0.5 * ws
    cw /= ws.sum()
    return float(np.interp(q, cw, xs))
````

````python
# ============================================================================
# AGREGADOS GEOGRÁFICOS
# ============================================================================
def agregados_geo(sdem: pd.DataFrame, por: list, prefijo: str) -> pd.DataFrame:
    d = sdem[sdem["c_res"].ne("2") & sdem["eda"].ge(15)].copy()
    d["w"] = d["fac_tri"]

    filas = []
    for llaves, g in d.groupby(por, dropna=False, observed=True):
        llaves = llaves if isinstance(llaves, tuple) else (llaves,)
        w   = g["w"]
        ocu = g[g["clase2"].eq(1)]
        pea = g[g["clase1"].eq(1)]
        asal = ocu[ocu["pos_ocu"].eq(1)]

        r = dict(zip(por, llaves))
        r.update({
            "n_personas_muestra":  len(g),
            "n_ocupados_muestra":  len(ocu),
            "pob_15mas_exp":       float(w.sum()),
            "pea_exp":             float(pea["w"].sum()),
            "ocupados_exp":        float(ocu["w"].sum()),
            "tasa_participacion":  float(pea["w"].sum() / w.sum()) if w.sum() else np.nan,
            "tasa_desocupacion":   float(g[g["clase2"].eq(2)]["w"].sum() / pea["w"].sum())
                                   if pea["w"].sum() else np.nan,
            "ing_medio":           media_pond(ocu["ingocup"], ocu["w"]),
            "ing_mediano":         cuantil_pond(ocu["ingocup"], ocu["w"], 0.50),
            "ing_p25":             cuantil_pond(ocu["ingocup"], ocu["w"], 0.25),
            "ing_p75":             cuantil_pond(ocu["ingocup"], ocu["w"], 0.75),
            "ing_p90":             cuantil_pond(ocu["ingocup"], ocu["w"], 0.90),
            "ing_hora_medio":      media_pond(ocu["ing_x_hrs"], ocu["w"]),
            "ing_medio_asalariado": media_pond(asal["ingocup"], asal["w"]),
            "hrs_medias":          media_pond(ocu["hrsocup"], ocu["w"]),
            "escolaridad_media":   media_pond(g["anios_esc"], w),
            "pct_informal":        media_pond(ocu["emp_ppal"].eq(1).astype(float), ocu["w"]),
            "pct_con_segsoc":      media_pond(ocu["seg_soc"].eq(1).astype(float), ocu["w"]),
            "pct_asalariado":      media_pond(ocu["pos_ocu"].eq(1).astype(float), ocu["w"]),
            "pct_cuenta_propia":   media_pond(ocu["pos_ocu"].eq(3).astype(float), ocu["w"]),
            "pct_medio_sup_o_mas": media_pond(g["niv_ins"].eq(4).astype(float), w),
            "edad_media":          media_pond(g["eda"], w),
        })
        filas.append(r)

    out = pd.DataFrame(filas)
    metricas = [c for c in out.columns if c not in por]
    return out.rename(columns={c: f"{prefijo}{c}" for c in metricas})


geo_ent  = agregados_geo(tip["sdem"], ["cve_ent"],                "ent_")
geo_muni = agregados_geo(tip["sdem"], ["cvegeo"],                 "mun_")
geo_cda  = agregados_geo(tip["sdem"], ["cd_a"],                   "cda_")
geo_ext  = agregados_geo(tip["sdem"], ["cve_ent", "t_loc_tri"],   "ext_")

for nombre, df in [("entidad", geo_ent), ("municipio", geo_muni),
                   ("ciudad autorrep.", geo_cda), ("entidad x tam. localidad", geo_ext)]:
    print(f"{nombre:>26s}: {len(df):>5,} celdas")
````

````python
# ============================================================================
# DIAGNÓSTICO DE TAMAÑO DE MUESTRA POR CELDA
# ============================================================================
diag = pd.DataFrame({
    "entidad":                  geo_ent["ent_n_personas_muestra"].describe(),
    "municipio":                geo_muni["mun_n_personas_muestra"].describe(),
    "ciudad_autorrep":          geo_cda["cda_n_personas_muestra"].describe(),
    "entidad_x_tam_localidad":  geo_ext["ext_n_personas_muestra"].describe(),
}).T
display(diag[["count", "mean", "min", "25%", "50%", "75%", "max"]])

n = geo_muni["mun_n_personas_muestra"]
print(f"\nMunicipios en la muestra: {len(n):,}")
for umbral in (30, 50, 100, 200, 500):
    print(f"  con menos de {umbral:>3} personas en muestra: "
          f"{(n < umbral).sum():>4,} ({100*(n < umbral).mean():5.1f}%)")
print("\n=> Confirma la advertencia: el corte municipal es frágil en la cola.")
````

### 8.1 Encogimiento del estimador municipal hacia la media estatal

Un promedio municipal con `n = 18` no se debe usar crudo. Aplico un encogimiento
tipo James–Stein / empírico bayesiano:

$$\hat{\theta}_m = \lambda_m \cdot \bar{y}_m + (1-\lambda_m)\cdot \bar{y}_{ent(m)},
\qquad \lambda_m = \frac{n_m}{n_m + k}$$

con `k` = tamaño de muestra al que se le da 50% de peso al dato local. Esto deja
la variable geográfica lista para entrar a un modelo sin inyectar ruido.

````python
# ============================================================================
# ENCOGIMIENTO MUNICIPIO -> ENTIDAD
# ============================================================================
K_SHRINK = 100  # n al que el municipio pesa 50%

def encoger(muni: pd.DataFrame, ent: pd.DataFrame, metricas: list,
            k: float = K_SHRINK) -> pd.DataFrame:
    m = muni.copy()
    m["cve_ent"] = m["cvegeo"].str[:2]
    m = m.merge(ent, on="cve_ent", how="left")

    n = m["mun_n_personas_muestra"].astype(float)
    lam = n / (n + k)
    m["mun_lambda_shrink"] = lam

    for met in metricas:
        col_m, col_e = f"mun_{met}", f"ent_{met}"
        if col_m in m and col_e in m:
            m[f"mun_{met}_shrunk"] = lam * m[col_m].fillna(m[col_e]) + (1 - lam) * m[col_e]
    return m


METRICAS_SHRINK = ["ing_medio", "ing_mediano", "ing_hora_medio", "escolaridad_media",
                   "pct_informal", "pct_con_segsoc", "pct_asalariado",
                   "pct_medio_sup_o_mas", "tasa_desocupacion", "tasa_participacion"]

geo_muni_sh = encoger(geo_muni, geo_ent, METRICAS_SHRINK)

display(geo_muni_sh[["cvegeo", "mun_n_personas_muestra", "mun_lambda_shrink",
                     "mun_ing_mediano", "ent_ing_mediano", "mun_ing_mediano_shrunk"]]
        .sort_values("mun_n_personas_muestra")
        .head(10))
````

## 10. Validación: reconstruir los indicadores oficiales

Si el pipeline no rompió nada, los agregados ponderados deben reproducir las cifras que
publica INEGI para el trimestre. Esta celda es la prueba de aceptación real.

````python
# ============================================================================
# INDICADORES OFICIALES ENOE
# ============================================================================
u = unificado
adulto = u["eda"].ge(15) & u["c_res"].ne("2")
w = u["fac_tri"]

pob15   = w[adulto].sum()
pea     = w[adulto & u["clase1"].eq(1)].sum()
ocu     = w[adulto & u["clase2"].eq(1)].sum()
des     = w[adulto & u["clase2"].eq(2)].sum()
info    = w[adulto & u["clase2"].eq(1) & u["emp_ppal"].eq(1)].sum()
subocu  = w[adulto & u["sub_o"].eq(1)].sum()

ind = pd.DataFrame([
    ("Población de 15 años y más",          pob15,          ""),
    ("Población económicamente activa",     pea,            ""),
    ("Población ocupada",                   ocu,            ""),
    ("Población desocupada",                des,            ""),
    ("Tasa de participación (%)",           100*pea/pob15,  "PEA / pob 15+"),
    ("Tasa de desocupación (%)",            100*des/pea,    "desocupados / PEA"),
    ("Tasa de informalidad laboral TIL1 (%)", 100*info/ocu, "empleo informal / ocupados"),
    ("Tasa de subocupación (%)",            100*subocu/ocu, "subocupados / ocupados"),
], columns=["indicador", "valor", "definicion"])
display(ind)

ocupados = u[adulto & u["clase2"].eq(1)]
print(f"\nIngreso mensual de ocupados (pesos corrientes):")
print(f"  media ponderada    : {media_pond(ocupados['ingocup'], ocupados['fac_tri']):>12,.0f}")
print(f"  mediana ponderada  : {cuantil_pond(ocupados['ingocup'], ocupados['fac_tri']):>12,.0f}")
print(f"  % ocupados sin dato: {100*ocupados['ingocup'].isna().mean():>12.2f}")
print("\nCompara contra el comunicado de INEGI del trimestre para dar por buena la corrida.")
````

````python
# ============================================================================
# CONTROL DE CALIDAD DEL MERGE
# ============================================================================
qc = (u[[c for c in u.columns if c.startswith("_m_")]]
        .mean().mul(100).round(2).rename("% filas con match").to_frame())
display(qc)

print("\nNulos en variables clave del modelo de ingreso:")
clave = ["ingocup", "hrsocup", "anios_esc", "eda", "sex", "niv_ins", "pos_ocu",
         "emp_ppal", "seg_soc", "scian", "hog_ing_pc", "mun_ing_mediano_shrunk"]
display((u[[c for c in clave if c in u.columns]].isna().mean() * 100)
        .round(2).rename("% nulos").to_frame())
````

````python
# ============================================================================
# OPTIMIZACIÓN DE MEMORIA
# ============================================================================
# Con los datos reales SDEM ronda las 250 mil personas y el unificado supera las
# 550 columnas. Casi todas las columnas de código son texto de baja cardinalidad:
# pasarlas a `category` recorta la memoria en un orden de magnitud sin perder nada.

def compactar(df: pd.DataFrame, umbral: float = 0.5) -> pd.DataFrame:
    out = df.copy()
    antes = out.memory_usage(deep=True).sum() / 1024**2
    for c in out.columns:
        s = out[c]
        if s.dtype.kind in "if":
            out[c] = pd.to_numeric(s, downcast="float" if s.dtype.kind == "f" else "integer")
        elif str(s.dtype) in ("object", "string"):
            if s.nunique(dropna=True) / max(len(s), 1) < umbral:
                out[c] = s.astype("category")
    despues = out.memory_usage(deep=True).sum() / 1024**2
    print(f"  {antes:,.0f} MB -> {despues:,.0f} MB  ({100*(1 - despues/antes):.0f}% menos)")
    return out


print("Compactando `unificado`:")
unificado = compactar(unificado)
````

````python
# ============================================================================
# TABLA DE ENRIQUECIMIENTO GEOGRÁFICO (llave CVEGEO municipal)
# ============================================================================
tabla_geo = (geo_muni_sh
             .assign(cve_ent=lambda d: d["cvegeo"].str[:2],
                     cve_mun=lambda d: d["cvegeo"].str[2:])
             .drop_duplicates("cvegeo"))

cols_geo = (["cvegeo", "cve_ent", "cve_mun", "mun_n_personas_muestra", "mun_lambda_shrink"]
            + [c for c in tabla_geo.columns if c.endswith("_shrunk")]
            + [c for c in tabla_geo.columns if c.startswith("ent_")])
tabla_geo = tabla_geo[[c for c in dict.fromkeys(cols_geo) if c in tabla_geo.columns]]

print(f"tabla_geo: {tabla_geo.shape[0]:,} municipios x {tabla_geo.shape[1]:,} columnas")
print("Llave 'cvegeo' (2 dígitos entidad + 3 municipio) = misma llave que el IRS de CONEVAL.")
display(tabla_geo.head())
````

## 12. Diccionario del dataset final y exportación

````python
# ============================================================================
# DICCIONARIO DEL DATASET UNIFICADO
# ============================================================================
desc = (dicc.dropna(subset=["descripcion"])
            .drop_duplicates("variable")
            .set_index("variable")["descripcion"].to_dict())

DESC_DERIVADAS = {
    "hog_n_personas": "Personas residentes habituales en el hogar",
    "hog_n_perceptores": "Personas del hogar con ingreso por trabajo > 0",
    "hog_ing_total": "Suma de ingocup de todos los residentes del hogar",
    "hog_ing_pc": "Ingreso total del hogar entre número de personas",
    "hog_ing_x_perceptor": "Ingreso total del hogar entre número de perceptores",
    "hog_razon_dep": "(menores de 15 + 65 y más) / personas de 15 a 64",
    "hog_pct_informal": "Proporción de ocupados del hogar en empleo informal",
    "hog_pct_segsoc": "Proporción de ocupados del hogar con acceso a seguridad social",
    "mun_lambda_shrink": "Peso del dato municipal en el encogimiento hacia la entidad",
}


def describir(col: str) -> str:
    if col in desc:
        return desc[col]
    if col in DESC_DERIVADAS:
        return DESC_DERIVADAS[col]
    for pref, nota in [("c1_", "COE1 · "), ("c2_", "COE2 · "), ("hg_", "HOG · "),
                       ("vv_", "VIV · "), ("jefe_", "Jefe de hogar · ")]:
        if col.startswith(pref):
            return nota + desc.get(col[len(pref):], "(derivada)")
    for pref, nota in [("mun_", "Agregado municipal ponderado"),
                       ("ent_", "Agregado estatal ponderado"),
                       ("cda_", "Agregado por ciudad autorrepresentada"),
                       ("ext_", "Agregado entidad x tamaño de localidad"),
                       ("hog_", "Agregado del hogar"),
                       ("_m_",  "Indicador de match del merge")]:
        if col.startswith(pref):
            return nota
    return "(sin descripción en el diccionario de INEGI)"


diccionario_final = pd.DataFrame({
    "variable": unificado.columns,
    "nivel": [nivel_de(c) for c in unificado.columns],
    "dtype": [str(unificado[c].dtype) for c in unificado.columns],
    "pct_nulos": [round(100 * unificado[c].isna().mean(), 2) for c in unificado.columns],
    "n_unicos": [unificado[c].nunique(dropna=True) for c in unificado.columns],
    "descripcion": [describir(c) for c in unificado.columns],
}).sort_values(["nivel", "variable"]).reset_index(drop=True)

display(diccionario_final.head(25))
print(f"\n{len(diccionario_final):,} variables documentadas")
````

````python
# ============================================================================
# EXPORTACIÓN
# ============================================================================
def exportar(df: pd.DataFrame, nombre: str):
    p = SALIDA / f"{nombre}.parquet"
    try:
        df.to_parquet(p, index=False)
        print(f"  {p}  ({p.stat().st_size/1024**2:6.1f} MB)")
    except Exception as e:
        p = SALIDA / f"{nombre}.csv.gz"
        df.to_csv(p, index=False, compression="gzip")
        print(f"  {p}  (parquet no disponible: {e})")


print("Exportando a", SALIDA.resolve())
exportar(unificado,           f"enoe_{TRIMESTRE}_unificado_persona")
exportar(tabla_modelo,        f"enoe_{TRIMESTRE}_tabla_modelo_ingreso")
exportar(hogar,               f"enoe_{TRIMESTRE}_agregados_hogar")
exportar(tabla_geo,           f"enoe_{TRIMESTRE}_enriquecimiento_municipal")
exportar(geo_ent,             f"enoe_{TRIMESTRE}_agregados_entidad")

diccionario_final.to_csv(SALIDA / f"enoe_{TRIMESTRE}_diccionario_unificado.csv",
                         index=False, encoding="utf-8-sig")
unicidad.to_csv(SALIDA / f"enoe_{TRIMESTRE}_verificacion_llaves.csv",
                index=False, encoding="utf-8-sig")
ref.to_csv(SALIDA / f"enoe_{TRIMESTRE}_integridad_referencial.csv",
           index=False, encoding="utf-8-sig")
print("\nListo.")
````

## 13. Notas metodológicas para el caso de uso

**1. La ENOE no se vincula a nivel persona con datos del buró.** No hay CURP ni RFC ni
identificador común. El uso legítimo es a nivel agregado (`tabla_geo`) o como corpus de
entrenamiento de un modelo estructural de ingreso que después se aplica a los atributos
que sí tienes del solicitante. Cualquier intento de reidentificación por combinación de
cuasi-identificadores rompería los términos de uso de los microdatos y el marco de la LFPDPPP.

**2. Los ponderadores no son opcionales.** La muestra de la ENOE está estratificada y
conglomerada, con sobremuestreo urbano. Un modelo ajustado sin `fac_tri` estima la
relación en la *muestra*, no en la población. Para regresión, `fac_tri` va como
`sample_weight`. Para errores estándar correctos hace falta además el diseño
(`est_d_tri` como estrato, `upm` como conglomerado); `statsmodels` no lo trae, pero
`samplics` sí, o se hace bootstrap por UPM.

**3. `ingocup` está truncado y censurado a la vez.** No se observa para desocupados ni
inactivos (censura por selección — el problema de Heckman que ya estás trabajando), y
tiene sub-declaración conocida en la cola alta. Además ~10-12% de los ocupados no declaran
monto y INEGI imputa vía `p6c` (múltiplos de salario mínimo). Vale la pena conservar
`c2_p6b2` y `c2_p6c` crudos para poder tratar los imputados aparte.

**4. Sesgo de selección respecto de tu población objetivo.** El universo de la ENOE es la
población en viviendas particulares; el de tu modelo es la población que *solicita crédito*.
La segunda es un subconjunto no aleatorio de la primera. Un modelo entrenado en ENOE y
aplicado a solicitantes tiene *covariate shift* garantizado. Si vas a usarlo así, conviene
reponderar la ENOE hacia el perfil demográfico de tus solicitantes (raking / IPW sobre
edad × sexo × entidad × escolaridad) antes de entrenar.

**5. Un trimestre es una foto.** `n_ent` indica la visita del panel rotativo (1 a 5). Si
más adelante quieres una estructura longitudinal, la ENOE es panel rotativo: la misma
vivienda se visita 5 trimestres. La llave `K_VIV` sin `n_ent`, `per` ni `mes_cal` permite
enlazar trimestres consecutivos, con las advertencias usuales sobre attrition y sobre
hogares mudados (`h_mud > 0`).

**6. Sobre el corte municipal.** Está construido y encogido, pero úsalo con el
`mun_lambda_shrink` a la vista. Para los municipios de la cola, el valor efectivo es casi
la media estatal, y eso está bien: es lo honesto. Si necesitas granularidad municipal real
y estable, el camino es combinar ENOE (señal laboral, trimestral) con Censo 2020 o la
ENIGH y con el IRS de CONEVAL, que sí son municipales por diseño.
